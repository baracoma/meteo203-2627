## Part 1: Generating a Single Wavelength Wave

**Task**: Use the code below to generate and plot a sine wave representing a single wave.

```python
import numpy as np
import matplotlib.pyplot as plt

# Define parameters
A = 1  # Amplitude
f = 5  # Frequency
time_start = 0
time_end = 1
t = np.linspace(time_start, time_end, 500)  # Time array (0 to 2pi)

# Generate the sine wave for a single wavelength
wave = A * np.sin(2 * np.pi * f * t)

# Plot the wave
plt.plot(t, wave)
plt.title("Single Wavelength Wave")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

## Part 3: Combining Multiple Waves

Now, let's combine multiple sine waves (representing different wavelengths).

**Task**: Run the following code to generate the combined wave.

```python
# Define multiple frequencies
freq1 = 5        # try a number between 1-5
freq2 = 20       # try a number between 5-20
freqs = [freq1, freq2] # Arbitrary frequencies representing different waves

time_start = 0
time_end = 1
t = np.linspace(time_start, time_end, 500)  # Time array (0 to 2pi)

# Generate the sum of waves
combined_wave = np.zeros_like(t)
for f in freqs:
    combined_wave += A * np.sin(2 * np.pi * f * t)

# Plot the combined wave
plt.plot(t, combined_wave)
plt.title("Sum of Multiple Wavelengths (Reduced Time)")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

## Part 4: Fourier Analysis of the Combined Wave

Now we will perform a Fourier transform to decompose the combined wave into its frequency components.

**Task**: Run the Fourier transform code and observe the output.

```python
from scipy.fft import fft, fftfreq

# Perform Fourier Transform
N = len(t)
T = t[1] - t[0]  # Sampling interval
yf = fft(combined_wave)
xf = fftfreq(N, T)[:N//2]

# Plot the Fourier Transform (showing the frequency components)
plt.plot(xf, 2.0/N * np.abs(yf[:N//2]))
plt.title("Fourier Transform (Frequency Components, Reduced Time)")
plt.xlabel("Frequency")
plt.ylabel("Amplitude")
plt.grid(True)
plt.xlim([0, 30])
plt.show()
```

## Part 5: Exploring Sampling Frequency

A continuous signal can only be measured at discrete points in time. The **sampling frequency** $f_s$ tells us how many measurements are taken per second.

In this example, we will keep the signal frequency fixed at:

$$
f = 10 \text{ Hz}
$$

and compare the same signal using different sampling frequencies.

For now, keep the sampling frequencies **above 20 Hz**, so that the 10-Hz signal remains below the Nyquist frequency.

**Task:** Compare a finely sampled signal with a more coarsely sampled signal. Change `fs_coarse` and observe how the representation of the wave changes.

```python
# Part 5: Effect of sampling frequency

# Signal properties
f = 10          # signal frequency (Hz)
A = 1
duration = 1    # seconds

# Sampling frequencies
fs_fine = 1000
fs_coarse = 50   # Try changing this, but keep it above 20 Hz for now

# Time arrays
t_fine = np.arange(0, duration, 1/fs_fine)
t_coarse = np.arange(0, duration, 1/fs_coarse)

# Generate the SAME 10-Hz sine wave
wave_fine = A * np.sin(2 * np.pi * f * t_fine)
wave_coarse = A * np.sin(2 * np.pi * f * t_coarse)

# Plot
plt.figure(figsize=(10, 6))

# Fine sampling
plt.subplot(2, 1, 1)
plt.plot(t_fine, wave_fine)
plt.title(f"Fine Sampling ($f_s$ = {fs_fine} Hz)")
plt.ylabel("Amplitude")
plt.grid(True)

# Coarse sampling
plt.subplot(2, 1, 2)

# Plot the finely sampled signal faintly as a reference
plt.plot(t_fine, wave_fine, alpha=0.3, label="Reference signal")

# Plot the actual sampled points
plt.plot(t_coarse, wave_coarse, 'o-', label="Sampled signal")

plt.title(f"Coarse Sampling ($f_s$ = {fs_coarse} Hz)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()

plt.suptitle(f"Sampling a {f}-Hz Signal")

plt.tight_layout()
plt.show()
```

## Part 6: Nyquist Frequency and Aliasing

We will now examine what happens when the sampling frequency becomes too low to represent the original signal correctly.

The **Nyquist frequency** is:

$$
f_N = \frac{f_s}{2}
$$

To represent a signal without aliasing, its highest frequency should be below the Nyquist frequency:

$$
f < f_N
$$

or equivalently,

$$
f_s > 2f
$$

For our 10-Hz signal, the sampling frequency should therefore be greater than 20 Hz.

In this exercise, we will compare:

- a very high sampling frequency,
- a sampling frequency just above the Nyquist requirement, and
- a sampling frequency below the Nyquist requirement.

When the sampling frequency is too low, **aliasing** occurs: the sampled data may appear to contain a different, lower-frequency signal.

```python
# Part 6: Nyquist frequency and aliasing

# Signal properties
f = 10          # signal frequency (Hz)
A = 1
duration = 1    # seconds

# Sampling frequencies
fs_reference = 1000   # very well sampled
fs_above = 25         # just above the Nyquist requirement
fs_below = 15         # below the Nyquist requirement

# Time arrays
t_reference = np.arange(0, duration, 1/fs_reference)
t_above = np.arange(0, duration, 1/fs_above)
t_below = np.arange(0, duration, 1/fs_below)

# Generate the SAME 10-Hz sine wave
wave_reference = A * np.sin(2 * np.pi * f * t_reference)
wave_above = A * np.sin(2 * np.pi * f * t_above)
wave_below = A * np.sin(2 * np.pi * f * t_below)

# Plot
plt.figure(figsize=(10, 8))

# Reference signal
plt.subplot(3, 1, 1)
plt.plot(t_reference, wave_reference)
plt.title(f"Reference Signal ($f_s$ = {fs_reference} Hz)")
plt.ylabel("Amplitude")
plt.grid(True)

# Just above Nyquist requirement
plt.subplot(3, 1, 2)
plt.plot(t_reference, wave_reference, alpha=0.3, label="Reference signal")
plt.plot(t_above, wave_above, 'o-', label="Sampled signal")
plt.title(f"Above Nyquist Requirement ($f_s$ = {fs_above} Hz)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()

# Below Nyquist requirement
plt.subplot(3, 1, 3)
plt.plot(t_reference, wave_reference, alpha=0.3, label="Reference signal")
plt.plot(t_below, wave_below, 'o-', label="Sampled signal")
plt.title(f"Below Nyquist Requirement ($f_s$ = {fs_below} Hz)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.legend()

plt.suptitle(f"Nyquist Sampling and Aliasing for a {f}-Hz Signal")

plt.tight_layout()
plt.show()
```
