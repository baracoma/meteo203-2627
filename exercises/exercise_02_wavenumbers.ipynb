## Part 1: Wavelength and Wavenumber

We will begin with a simple sinusoidal temperature pattern across a 100-meter domain.

The wavelength $\lambda$ describes the distance over which the pattern completes one full cycle. The corresponding angular wavenumber is:

$$
k = \frac{2\pi}{\lambda}
$$

A **long wavelength** corresponds to a **small wavenumber**, while a **short wavelength** corresponds to a **large wavenumber**.

**Task:** In the code below, set `wavelength` to **20 meters** and run the cell. Then try other wavelengths, such as 10 m and 40 m, and observe how the spatial pattern changes. Return `wavelength` to **20 m** before proceeding to Part 2.


```python
import numpy as np
import matplotlib.pyplot as plt

# Define the length of the domain
L = 100  # meters

# Temperature signal properties
mean_temperature = 10  # °C
amplitude = 5           # °C

# EDIT HERE: start with a wavelength of 20 meters
wavelength = 20  # meters

# Calculate the wavenumber
k = 2 * np.pi / wavelength  # radians per meter

# Generate positions across the domain at fine resolution
x_fine = np.linspace(0, L, 1000)

# Generate the spatial temperature variation
temperature = mean_temperature + amplitude * np.sin(k * x_fine)

# Plot
plt.figure(figsize=(10, 4))
plt.plot(x_fine, temperature)
plt.title(
    f"Temperature Variation: wavelength = {wavelength} m, "
    f"k = {k:.3f} rad/m"
)
plt.xlabel("Position (m)")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.show()
```

## Part 3: Fundamental and Nyquist Wavenumbers

Two useful limits describe the range of spatial scales represented by a dataset.

### Fundamental wavenumber

For a domain of length $L$, the **fundamental wavenumber** is:

$$
k_0 = \frac{2\pi}{L}
$$

It is the smallest non-zero wavenumber represented by the domain and corresponds to the longest wavelength that fits once across the domain:

$$
\lambda_{\max} = L
$$

### Nyquist wavenumber

For measurements separated by $\Delta x$, the **Nyquist wavenumber** is:

$$
k_N = \frac{\pi}{\Delta x}
$$

It is the largest wavenumber that can be represented without spatial aliasing under ideal sampling. The corresponding shortest wavelength is:

$$
\lambda_N = 2\Delta x
$$

## Part 4: Spatial Aliasing

We will now deliberately sample the same 20-m temperature wave at two different spatial resolutions.

For a wavelength of 20 m, the spatial sampling interval should be smaller than half the wavelength:

$$
\Delta x < \frac{\lambda}{2}
$$

For this example:

$$
\frac{\lambda}{2} = 10 \text{ m}
$$

We will compare:

- **5-m spacing**, which provides four samples per wavelength; and
- **15-m spacing**, which is too coarse to represent the 20-m wave correctly.

When the spacing is too large, the sampled measurements can suggest a different spatial pattern. This is **spatial aliasing**.


```python
# Baseline spatial signal
wavelength = 20  # meters
k = 2 * np.pi / wavelength

x_fine = np.linspace(0, L, 1000)
temperature = mean_temperature + amplitude * np.sin(k * x_fine)

# Two sampling intervals
spacing_adequate = 5   # meters
spacing_aliased = 15   # meters

# Sensor locations
x_adequate = np.arange(0, L + 1e-9, spacing_adequate)
x_aliased = np.arange(0, L + 1e-9, spacing_aliased)

# Sample the same underlying temperature field
temp_adequate = mean_temperature + amplitude * np.sin(k * x_adequate)
temp_aliased = mean_temperature + amplitude * np.sin(k * x_aliased)

# Plot
plt.figure(figsize=(10, 9))

# Reference signal
plt.subplot(3, 1, 1)
plt.plot(x_fine, temperature)
plt.title(f"Reference Spatial Signal (wavelength = {wavelength} m)")
plt.ylabel("Temperature (°C)")
plt.grid(True)

# Adequate sampling
plt.subplot(3, 1, 2)
plt.plot(x_fine, temperature, alpha=0.30, label="Reference signal")
plt.plot(x_adequate, temp_adequate, "o-", label="Sensor measurements")
plt.title(f"Adequate Sampling (sensor spacing = {spacing_adequate} m)")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid(True)

# Aliased sampling
plt.subplot(3, 1, 3)
plt.plot(x_fine, temperature, alpha=0.30, label="Reference signal")
plt.plot(x_aliased, temp_aliased, "o-", label="Sensor measurements")
plt.title(f"Undersampling and Aliasing (sensor spacing = {spacing_aliased} m)")
plt.xlabel("Position (m)")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
